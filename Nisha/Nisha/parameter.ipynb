# parameter vs Arguments

```python
def stu(name): # parameter
    print(name)
stu("shabnam") #  arguments
```

# 1 Positional Arguments

```python
def student(name,age):
    print("name:", name)
    print("agr: ",age)
student("shabnam",23)
```

# /-Positional - only

```python
def s(name,age,/):
    print(name,age)
```

```python
s("kirti",23)
```

# /- Keywords- only

```python
def student(*name):
    print(name)
student("kirti","23","45")
```
