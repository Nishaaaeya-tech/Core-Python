# 2
# Below ₹500 → No discount
# 500–₹999 → 5%
# ₹1000–₹1999 → 10%
# ₹2000 or more → 15%
# write a Python program to calculate the discount and final price.

```python
amount= float(input("Enter your price:$ "))
if amount<500:
    discount=0
elif amount<=900:
    discount=5
elif amount<=1999:
    discount=10
else:
    discount=15
discount_amount=amount*discount/100
final_price=amount-discount_amount
print("discount:",discount,"%")
print("Discount Amount: ₹", discount_amount)
print("final_price:$",final_price)
```
