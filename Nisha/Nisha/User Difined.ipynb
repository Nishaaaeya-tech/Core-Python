```python
def cal(a,b,operator):
    if operator=="+":
        print("your output is :",a+b)
    elif operator=="-":
        print("your output is :",a-b)
    elif operator=="*":
        print("your output is :",a*b)
    elif operator=="/":
        print("your output is :",a/b)
    else:
        print("incorrect value")
a=int(input("enter 1st number: "))
b=int(input("enter 2nd number: "))
operator=input("Enter operator(+,-,*,/): ")
cal(a,b,operator)
```
